```python
import pandas as pd

bls = pd.read_excel("national_owner_M2024_dl.xlsx")
ai = pd.read_excel("AIOE_DataAppendix.xlsx", sheet_name="Appendix A")

print("=== BLS Spalten ===")
print(bls.columns.tolist())

print("\n=== Felten Spalten ===")
print(ai.columns.tolist())

print("\n=== BLS erste Zeilen ===")
print(bls.head(3))

print("\n=== Felten erste Zeilen ===")
print(ai.head(3))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# ── 1. FILTER & MERGE ──────────────────────────────────────────────────────────

# Keep only cross-industry, all ownership rows (OWN_CODE 1235 = all)
# and detailed occupations only (OCC_CODE ends in digits, not X000)
bls_filtered = bls[
    (bls["OWN_CODE"] == 1235) &
    (bls["O_GROUP"] == "detailed")
].copy()

# Clean SOC Code column in Felten (remove spaces just in case)
ai["SOC Code"] = ai["SOC Code"].str.strip()
bls_filtered["OCC_CODE"] = bls_filtered["OCC_CODE"].str.strip()

# Merge on SOC code
df = pd.merge(
    ai,
    bls_filtered[["OCC_CODE", "OCC_TITLE", "A_MEDIAN", "TOT_EMP"]],
    left_on="SOC Code",
    right_on="OCC_CODE"
)

# Convert to numeric (BLS stores some values as strings like "*")
df["A_MEDIAN"] = pd.to_numeric(df["A_MEDIAN"], errors="coerce")
df["TOT_EMP"]  = pd.to_numeric(df["TOT_EMP"],  errors="coerce")
df = df.dropna(subset=["A_MEDIAN", "AIOE"])

print(f"Merged dataset: {len(df)} occupations")
print(df[["Occupation Title", "AIOE", "A_MEDIAN"]].head(10))
```

```python
bls_filtered = bls[
    (bls["OWN_CODE"].astype(str) == "123") &
    (bls["O_GROUP"] == "detailed")
].copy()

print("BLS filtered rows:", len(bls_filtered))
print(bls_filtered["OCC_CODE"].head(10).tolist())
```

```python
# Clean and merge
ai["SOC Code"] = ai["SOC Code"].str.strip()
bls_filtered["OCC_CODE"] = bls_filtered["OCC_CODE"].str.strip()

df = pd.merge(
    ai,
    bls_filtered[["OCC_CODE", "OCC_TITLE", "A_MEDIAN", "TOT_EMP"]],
    left_on="SOC Code",
    right_on="OCC_CODE"
)

# Convert to numeric
df["A_MEDIAN"] = pd.to_numeric(df["A_MEDIAN"], errors="coerce")
df["TOT_EMP"]  = pd.to_numeric(df["TOT_EMP"],  errors="coerce")
df = df.dropna(subset=["A_MEDIAN", "AIOE"])

print(f"Merged dataset: {len(df)} occupations")
print(df[["Occupation Title", "AIOE", "A_MEDIAN"]].head(10))
```

```python
# ── 2. QUARTILE GROUPS ─────────────────────────────────────────────────────────

df["AI_Quartile"] = pd.qcut(df["AIOE"], q=4,
                             labels=["Q1 (Low)", "Q2", "Q3", "Q4 (High)"])

# ── 3. PLOT SETTINGS ───────────────────────────────────────────────────────────

sns.set_theme(style="whitegrid", font_scale=1.2)
fig, axes = plt.subplots(2, 2, figsize=(14, 11))
fig.suptitle("AI Exposure and Wages – Descriptive Analysis", fontsize=16, y=1.01)

# ── PLOT 1: Boxplot – Median Wage by AI Quartile ───────────────────────────────

sns.boxplot(data=df, x="AI_Quartile", y="A_MEDIAN", ax=axes[0,0],
            palette="Blues")
axes[0,0].set_title("Median Annual Wage by AI Exposure Quartile")
axes[0,0].set_xlabel("AI Exposure Quartile")
axes[0,0].set_ylabel("Median Annual Wage (USD)")

# ── PLOT 2: Scatterplot – AIOE vs Wage ─────────────────────────────────────────

axes[0,1].scatter(df["AIOE"], df["A_MEDIAN"], alpha=0.4, color="steelblue", s=20)
m, b = np.polyfit(df["AIOE"], df["A_MEDIAN"], 1)
x_line = np.linspace(df["AIOE"].min(), df["AIOE"].max(), 100)
axes[0,1].plot(x_line, m * x_line + b, color="tomato", linewidth=2)
axes[0,1].set_title("AI Exposure vs. Median Annual Wage")
axes[0,1].set_xlabel("AI Occupational Exposure (AIOE)")
axes[0,1].set_ylabel("Median Annual Wage (USD)")

# ── PLOT 3: Bar – Average Wage by Quartile ─────────────────────────────────────

avg_wage = df.groupby("AI_Quartile", observed=True)["A_MEDIAN"].mean().reset_index()
sns.barplot(data=avg_wage, x="AI_Quartile", y="A_MEDIAN", ax=axes[1,0],
            palette="Blues")
axes[1,0].set_title("Average Wage by AI Exposure Quartile")
axes[1,0].set_xlabel("AI Exposure Quartile")
axes[1,0].set_ylabel("Average Annual Wage (USD)")

# ── PLOT 4: Top 10 vs Bottom 10 Occupations ────────────────────────────────────

top10    = df.nlargest(10,  "AIOE")[["Occupation Title", "AIOE", "A_MEDIAN"]]
bottom10 = df.nsmallest(10, "AIOE")[["Occupation Title", "AIOE", "A_MEDIAN"]]
compare  = pd.concat([
    top10.assign(Group="High AI Exposure"),
    bottom10.assign(Group="Low AI Exposure")
])

sns.barplot(data=compare, x="A_MEDIAN", y="Occupation Title",
            hue="Group", ax=axes[1,1],
            palette={"High AI Exposure": "steelblue", "Low AI Exposure": "tomato"})
axes[1,1].set_title("Top 10 vs Bottom 10 AI-Exposed Occupations")
axes[1,1].set_xlabel("Median Annual Wage (USD)")
axes[1,1].set_ylabel("")

plt.tight_layout()
plt.savefig("ai_wage_analysis.png", dpi=150, bbox_inches="tight")
plt.show()

print("\n=== Summary Statistics by Quartile ===")
print(df.groupby("AI_Quartile", observed=True)["A_MEDIAN"].describe().round(0))
```
